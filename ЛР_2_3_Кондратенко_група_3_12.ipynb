```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
source = 'https://uk.wikipedia.org/wiki/%D0%9D%D0%B0%D1%81%D0%B5%D0%BB%D0%B5%D0%BD%D0%BD%D1%8F_%D0%A3%D0%BA%D1%80%D0%B0%D1%97%D0%BD%D0%B8#%D0%9D%D0%B0%D1%80%D0%BE%D0%B4%D0%B6%D1%83%D0%B2%D0%B0%D0%BD%D1%96%D1%81%D1%82%D1%8C'
df = pd.read_html(source, match='Коефіцієнт народжуваності в регіонах України', thousands=".", decimal=",", storage_options={"User-Agent": "Mozilla/5.0"})[0]
df.head(5)
```

```python
lines, columns = df.shape
print(f'Кількість рядків: {lines}')
print(f'Кількість стовбців: {columns}')
```

```python
# Заміняємо текстові пропуски
df = df.replace({'—': np.nan})
df.head(5)
```

```python
df.dtypes
```

```python
nan_columns = df.columns[df.isnull().any()].tolist()
df[nan_columns] = df[nan_columns].astype(float)

df.dtypes
```

```python
# Розрахунок частки пропущених значень
df.isnull().sum() / len(df)
```

```python
df = df.drop(27, errors='ignore')
df.tail(5)
```

```python
# Заповняємо пропуски середніми значеннями стовпців
df = df.fillna(df.mean(numeric_only=True), axis=0)
df.head(5)
```

```python
#Фільтрація регіонів із показниками вище середнього
avg = df['2019'].mean()
df.loc[df['2019'] > avg] = df['2019'].mean()
df.loc[df['2019'] > avg]
```

```python
#Пошук регіону з максимальним показником за 2014 рік.
max_value_region = df[df['2014'] == df['2014'].max()]['Регіон']
max_value_region
```

```python
years = ['1950', '1960', '1970', '1990', '2000', '2012', '2014', '2019']
mean_values = df[years].mean()

plt.figure(figsize=(10, 5))
plt.plot(years, mean_values, marker='o', color='g', linewidth=2)

plt.title('Динаміка середнього коефіцієнта народжуваності в Україні')
plt.xlabel('Рік')
plt.ylabel('Середній коефіцієнт')
plt.grid(True)
plt.show()
```

```python
# 1. Приводимо колонку '2019' до числового типу
df['2019'] = pd.to_numeric(df['2019'], errors='coerce')

# 2. Видаляємо рядки з пропущеними значеннями (NaN) у назвах регіонів та показниках
df_clean = df.dropna(subset=['Регіон', '2019']).copy()

# 3. Примусово перетворюємо назви регіонів на строк (str)
df_clean['Регіон'] = df_clean['Регіон'].astype(str)

# 4. Сортуємо дані
df_sorted = df_clean.sort_values(by='2019', ascending=True)

plt.figure(figsize=(10, 8))
plt.barh(df_sorted['Регіон'], df_sorted['2019'], color='black', edgecolor='black')
plt.title('Коефіцієнт народжуваності за регіонами у 2019 році')
plt.xlabel('Коефіцієнт')
plt.ylabel('Регіон')
plt.grid(axis='x', alpha=0.8)
plt.tight_layout()
plt.show()
```

```python
# 1. Приводимо дані до числового типу та видаляємо порожні значення (NaN)
data_1990 = pd.to_numeric(df['1990'], errors='coerce').dropna()
data_2019 = pd.to_numeric(df['2019'], errors='coerce').dropna()

# 2. Побудова гістограми
plt.figure(figsize=(10, 5))

plt.hist(data_1990, bins=12, alpha=0.6, label='1990 рік', color='orange', edgecolor='black')
plt.hist(data_2019, bins=12, alpha=0.6, label='2019 рік', color='blue', edgecolor='black')

plt.title('Гістограма розподілу коефіцієнтів (1990 vs 2019)')
plt.xlabel('Коефіцієнт народжуваності')
plt.ylabel('Кількість регіонів')
plt.legend()
plt.grid(axis='y', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
years = ['1950', '1960', '1970', '1990', '2000', '2012', '2014', '2019']

# Знаходимо рядок, який містить слово "Київ" (наприклад, "м. Київ")
kyiv_row = df[df['Регіон'].str.contains('Київ', na=False)]

# Беремо значення для років
kyiv_data = kyiv_row[years].values[0]

plt.figure(figsize=(10, 5))
plt.step(years, kyiv_data, where='mid', color='blue', linewidth=2, marker='o')

plt.title('Динаміка народжуваності у Києві')
plt.xlabel('Рік')
plt.ylabel('Коефіцієнт народжуваності')
plt.grid(True)
plt.show()
```

```python
df['2019'] = pd.to_numeric(df['2019'], errors='coerce')

# 2. Знаходимо та видаляємо рядки, де у стовпчику 'Регіон' опинилися числа
is_numeric_region = pd.to_numeric(df['Регіон'], errors='coerce').notna()
df_clean = df[~is_numeric_region].dropna(subset=['Регіон', '2019']).copy()

df_clean['Регіон'] = df_clean['Регіон'].astype(str)

# 3. Побудова графіка
plt.figure(figsize=(10, 5))

x_indices = range(len(df_clean))
plt.step(x_indices, df_clean['2019'], where='mid')

plt.title('Ступінчастий графік народжуваності')
plt.xlabel('Регіон')
plt.ylabel('Коефіцієнт')

plt.xticks(ticks=x_indices, labels=df_clean['Регіон'], rotation=90)

plt.tight_layout()
plt.show()
```
