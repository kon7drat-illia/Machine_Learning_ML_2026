### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

mu = np.mean(A, axis=0)
sigma = np.std(A, axis=0)

z = (A - mu) / sigma

print(z)
````

Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.

````python
# Код студента
A = np.array([
    [1, 2, 1],
    [2, 1, -1],
    [3, -1, 2]
])
b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)
print(x)

print(A @ x)
print(np.allclose(A @ x, b))
````
